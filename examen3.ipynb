```python
# General
import pandas as pd

# Visualizaciones
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = "notebook"
```

```python
df = pd.read_excel('data/BD_Nacionalidad.xlsx')
df.head()
```

```python
# Filter dataset
df = df[['Fecha', 'Valor']]

# Convert date column to datetime
df['Fecha'] = pd.to_datetime(df['Fecha'])

# Group values
df_clean = df.groupby('Fecha')['Valor'].sum().reset_index()
df_clean.head()
```

```python
# Visualizar la serie
fig = go.Figure()
fig.add_trace(go.Scatter(x=df_clean['Fecha'], y=df_clean['Valor'].squeeze(), mode='lines', name='Turistas'))
fig.update_layout(title=f'Turismo en México',
                  yaxis_title='Turistas')
fig.show()
```
